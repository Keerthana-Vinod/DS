```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

import re
import string
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
text = """
My name is Keerthana Vinod. I am from Kerala, India. I completed my schooling at my hometown.
I belong to a supportive family of four members. Currently, I am pursuing B.Tech in Computer
Science and Artificial Intelligence. My achievements include participating in hackathons,
seminars, and technical workshops. My hobbies are drawing, listening to music, and crafting.
My interests include Artificial Intelligence, Machine Learning, and Web Development.
My strengths are quick learning, creativity, and teamwork. My weaknesses are public speaking
and overthinking, but I am continuously working to improve them.
"""

print(text)
```

```python
lower_text = text.lower()

print(lower_text)
```

```python
no_punctuation = lower_text.translate(str.maketrans('', '', string.punctuation))

print(no_punctuation)
```

```python
no_numbers = re.sub(r'\d+', '', no_punctuation)

print(no_numbers)
```

```python
from nltk.tokenize import word_tokenize

tokens = word_tokenize(no_numbers)

print(tokens)
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered_words = [word for word in tokens if word not in stop_words]

print(filtered_words)
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print(stemmed_words)
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print(lemmatized_words)
```

```python
unique_words = list(set(lemmatized_words))

print(unique_words)
```

```python
from collections import Counter

frequency = Counter(lemmatized_words)

print(frequency)
```

```python
final_text = " ".join(lemmatized_words)

print(final_text)
```
